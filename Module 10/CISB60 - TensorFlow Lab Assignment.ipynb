#### Import datetime

```python
import datetime
```

#### load the fashion_mnist dataset and split it into train and test

```python
fashion_mnist = tf.keras.datasets.fashion_mnist
```

#### Normalize the data for x

```python
(X_train, y_train),(X_test, y_test) = fashion_mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0
```

#### Write a function  that creates a simple Keras model for classifying the images into 10 classes. <br> def create_model():
- Use the sequential model
- flatten the data 
- add a hidden layer with 64 neurons with activation function relu
- add the dropout technique to reduce model overfitting
 - tf.keras.layers.Dropout(0.2),
- add an output layer with 10 neurons and activation function SoftMax

```python
def create_model():
    return tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
  ])
```

When training with Keras's Model.fit(), adding the tf.keras.callbacks.TensorBoard callback ensures that logs are created and stored. Additionally, enable histogram computation every epoch with histogram_freq=1 (this is off by default)

Place the logs in a timestamped subdirectory to allow easy selection of different training runs.

#### Open a second notebook and execute the following code in that notebook.

%load_ext tensorboard

%tensorboard --logdir logs/fit


#### Create a model and compile using the optimizer adam, the loss "sparse_categorical_crossentropy", and metrics = accuracy.

```python
# Enter your code here:
model = create_model()
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
# Generate the logs for Tensorboard
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir=log_dir, histogram_freq=1)

# fit the model 
history = model.fit(x=X_train, y=y_train, epochs=10, validation_data=(X_test, y_test), callbacks=[tensorboard_callback])
```

#### Interview Questions

#### How does Deep Learning differ from Machine Learning? 

Machine Learning (ML) refers to algorithms that learn patters from data and make prediction or decisions based on that data.  ML models typically rely on human-engineered features (data pre-processed and transformed).  Deep Learning (DL) is a subset of ML that uses artificial neural networks (ANNs) with many layers (hence the term "deep").  DL models automatically learn hieracrhical features from the raw data itself without the need for manual feature extraction.

#### What is a Computational Graph?

A computational graph is a representation of mathematical expressions or operations as a directed graph.  In the context of ML and DL, it is a way of visualizing and organizing the flow of data through various operations, as well as how the gradients flow during backpropagation for training a model.

#### Why do we use the activation function?

The activation function is a critical component of neural networks in deep learning.  Its primary purpose is to introduce non-linearity into the network, enabling it to learn complex patterns and make more powerful predictions.  Without activation functions, a neural network would simply be a linear model, no matter how many layers it has.
